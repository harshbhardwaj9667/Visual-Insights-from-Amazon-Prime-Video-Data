```python
import pandas as pd
```

```python
import numpy as np
```

```python
df=pd.read_csv("amazon_prime_titles.csv")
```

```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
df.head()
```

```python
df.drop(['show_id'], axis=1 , inplace=True)       # removing columns
df.drop(['description'], axis=1 , inplace=True)
```

```python
df.head()
```

```python
# checking for duplicate
df.duplicated().sum()
```

```python
df.info()
```

```python
df.isna().sum()     # df.isnull().sum()
```

```python
# Filling null values
df['director'].fillna('Unavailable', inplace= True)
df['cast'].fillna('Unavailable', inplace= True)
df['country'].fillna('Unavailable', inplace= True)
df['date_added']= df['date_added'].ffill()     # forward fill(ffill)-- takes above value and puts it inplace of null value
df['rating']=df['rating'].fillna(df['rating'].mode()[0])
```

```python
df.isna().sum()
```

```python
df['date_added']=pd.to_datetime(df['date_added'])
```

```python
df.info()
```

```python
df.head()
```

```python
# Top 10 directors with most directed movies
filtered_directors = df[df['director'] != 'Unavailable']
top_directors = filtered_directors['director'].value_counts()[:10]
```

```python
top_directors
```

```python
plt.figure(figsize=(12,6))
sns.barplot(x=top_directors.values,y=top_directors.index, palette='plasma')
plt.title('Top 10 Prolific Directors (Excluding Unavailable)')
plt.xlabel('Number of Movies')
plt.ylabel('Director')
plt.show()
```

```python
filtered_cast = df[df['cast'] != 'Unavailable']
top_cast = filtered_cast['cast'].value_counts()[:10]
```

```python
top_cast
```

```python
plt.figure(figsize=(12,6))
sns.barplot(x=top_cast.values,y=top_cast.index, palette='plasma')
plt.title('Top 10 Prolific Actors (Excluding Unavailable)')
plt.xlabel('Number of Movies')
plt.ylabel('Actor')
# plt.xticks(rotation=90)
plt.show()
```

```python
# Count of each content type (Movies vs. TV show)
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='type', palette='viridis')
plt.title('Count of Movies vs TV Shows')
plt.show()
```

```python
plt.pie(df["type"].value_counts(), labels=df["type"].value_counts().index, autopct="%1.2f%%", explode=[0.2,0])
plt.title('Movies and TV Shows Ratio')
plt.legend()
plt.show()
```

```python
# Calculating the number of titles per country
filtered_country = df[df['country'] != 'Unavailable']
country_counts = filtered_country['country'].value_counts()
```

```python
# Getting the top 3 countries
top_countries = country_counts.nlargest(3).index
top_countries
```

```python
# Filtering the dataframe to include only the top 3 countries
df_top_countries = filtered_country[filtered_country['country'].isin(top_countries)]
df_top_countries
```

```python
# Plotting the number of titles released per top 3 countries
plt.figure(figsize=(10, 6))
sns.countplot(data=df_top_countries, x='country' , hue="type" , palette='viridis')
plt.title('Number of Titles Released per Top 3 Countries')
plt.show()
```

```python
year_counts = df['release_year'].value_counts()
```

```python
# Getting the top 10 release year
top_years = year_counts.nlargest(10).index    #2012,2013,2014
# Filtering the dataframe to include only the top 10 years
df_top_years = df[df['release_year'].isin(top_years)]
```

```python
# Plotting the number of titles released per top 10 Years
plt.figure(figsize=(10, 6))
sns.countplot(data=df_top_years, x='release_year' , hue="type" , palette='viridis', order=top_years)
plt.title('Number of Titles Released per Top 10 Years')
plt.show()
```

```python
# Distribution of genres
# Splitting the genres
df['Genres']= df['listed_in'].apply(lambda x: x.split(', '))
df['Genres']
```

```python
# Exploding the genres into seperate rows
genres_exploded = df.explode('Genres')
genres_exploded
```

```python
x=genres_exploded['Genres'].value_counts().index
plt.figure(figsize=(14,7))
sns.countplot(data=genres_exploded, y='Genres', palette='viridis',order=x)
plt.title('Distribution of Genres')
plt.show()
```

```python
df["duration_num"]=df["duration"].apply(lambda x: int(x.split()[0]))
```

```python
# Filter data for movies and tv shows seperately
df_movies = df[df['type'] == 'Movie'].copy()
df_tv_shows = df[df['type'] == 'TV Show'].copy()
plt.figure(figsize=(12, 12))
```

```python
# Subplot 1 : Top 10 longest Movie Duration
plt.subplot(2, 1, 1)
sns.barplot(data=df_movies.sort_values(by='duration_num', ascending=False).head(10), y='title',x='duration_num', palette="viridis")
plt.title('Top 10 Longest Movies Durations')
plt.xlabel('Duration (Minutes)')
plt.ylabel('Movie Title')
# Subplot 2 : Top 10 TV Shows with the most Seasons
plt.subplot(2, 1, 2)
sns.barplot(data=df_tv_shows.sort_values(by='duration_num', ascending=False).head(10), y='title',x= 'duration_num',palette="viridis")
plt.title('Top 10 TV Shows with the Most Seasons')
plt.xlabel('Number of Seasons')
plt.ylabel('TV Show Title')

plt.tight_layout()
plt.show()
```
